```python
product_category = {
    # Clothing
    "T-shirt": "Clothing",
    "Shirt": "Clothing",
    "Shorts": "Clothing",
    "Hoodie": "Clothing",
    "Pants": "Clothing",
    "Socks": "Clothing",
    "Jeans": "Clothing",
    "Blouse": "Clothing",
    "Skirt": "Clothing",
    "Sweater": "Clothing",
    "Dress": "Clothing",

    # Footwear
    "Shoes": "Footwear",
    "Sandals": "Footwear",
    "Sneakers": "Footwear",
    "Boots": "Footwear",

    # Electronics
    "Laptop": "Electronics",
    "Phone": "Electronics",
    "Headphones": "Electronics",
    "Watch": "Electronics",

    # Accessories
    "Sunglasses": "Accessories",
    "Gloves": "Accessories",
    "Jewelry": "Accessories",
    "Hat": "Accessories",
    "Handbag": "Accessories",
    "Backpack": "Accessories",
    "Belt": "Accessories",
    "Scarf": "Accessories",
    "Bag": "Accessories",

    # Outerwear
    "Coat": "Outerwear",
    "Jacket": "Outerwear"
}
```

```python
# correcting products in category
df["Category"] = df["Item Purchased"].map(product_category)
```

```python
df.head(5)
```

```python
df.groupby('Category')['Item Purchased'].unique()
```

```python
df.groupby('Category')['Size'].unique()
```

```python
# filing size values in there category 
invalid_size_categories = ['Accessories', 'Electronics','Footwear']

df.loc[df['Category'].isin(invalid_size_categories), 'Size'] = 'Not_Applicable'
```

```python
df.groupby('Category')['Size'].unique()
```

```python
# Get mode for Clothing category
clothing_mode = df[df['Category'] == 'Clothing']['Size'].mode()[0]

# Fill null values in Clothing category
df.loc[
    (df['Category'] == 'Clothing') & (df['Size'].isna()),
    'Size'
] = clothing_mode
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.loc[df['Category'] == 'Footwear', 'Size'] = 'Not_Available'
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.isnull().sum()
```

```python
df.groupby('Category')['Review Rating'].unique()
```

```python
# filing reivew with there category based mean 
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(
    lambda x: x.fillna(x.mean())
)
```

```python
df.isnull().sum()
```

```python
df.groupby('Category')['Review Rating'].unique()
```

```python
df['Review Rating'] = df['Review Rating'].round(2)
```

```python
df.groupby('Category')['Review Rating'].unique()
```

```python
df['Previous Purchases'] = df['Previous Purchases'].fillna(0)
```

```python
df.isnull().sum()
```

```python
df.groupby('Category')['Purchase Amount (USD)'].unique()
```

```python
# filing values of purchase amount with there respective category with median

df['Purchase Amount (USD)'] = df.groupby('Category')['Purchase Amount (USD)'].transform(
    lambda x: x.fillna(x.median())
)
```

```python
df.isnull().sum()
```

```python
# corrcting columns names 
import re

df.columns = [
    re.sub(r'[^a-z0-9_]', '', col.lower().replace(' ', '_'))
    for col in df.columns
]
```

```python
# renaming column 
df.rename(columns={"purchase_amount_usd ": "purchase_amount"}, inplace=True)
```
